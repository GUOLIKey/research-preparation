```python
import numpy as np
```

```python
rng = np.random.default_rng(42)
X = rng.normal(size=(1000, 20))
```

Create a random number generator with a fixed seed (42),
then generate a 1000x20 matrix from the standard normal distribution.

```python
means = X.mean(axis=0)
```

Compute the mean of each feature along axis=0.

```python
stds = X.std(axis=0)
```

Compute the standard deviation of each feature along axis=0.

```python
X_norm = (X - means) / stds
```

Normalization

```python
print(X_norm.mean(axis=0))
print(X_norm.std(axis=0))
```

Check the normalization result

```python
norms = np.linalg.norm(X, axis=1)
top5_indices = np.argsort(norms)[-5:][::-1]
```

Compute the Euclidean norm,
find the indices of the 5 samples with the largest norms.


```python
def cosine_similarity(x, y):
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))
```

Implement my own cosine similarity function

```python
X_first20 = X[:20]
norms20 = np.linalg.norm(X_first20, axis=1)
pairwise_sim = (X_first20 @ X_first20.T) / np.outer(norms20, norms20)
```

X @ X.T, multiplies the matrix by its own transpose.
np.outer(a, b) builds a matrix whose [i, j] entry is a[i] * b[j], a=norms20, b=norms20

```python
from sklearn.metrics.pairwise import cosine_similarity as skl_cos
```

```python
sk_sim = skl_cos(X_first20)
print(np.allclose(pairwise_sim, sk_sim))
print(cosine_similarity(X_first20[0], X_first20[1]))
print(sk_sim[0, 1])
```
